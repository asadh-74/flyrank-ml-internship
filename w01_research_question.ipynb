## 2. The research question

**Question:** Given a content page's recent search and engagement signals, should it be prioritized for review?

| Element | Answer |
|---|---|
| **Decision** | Which page should the content team review first? |
| **Unit of analysis** | One content page (one row per `content_id`) |
| **Action** | Refresh, expand, protect, prune, or monitor |
| **Cost of wrong "review" call** | Wasted reviewer time; missed opportunity on better candidates |
| **Cost of wrong "ignore" call** | Declining page keeps losing traffic unchecked |
| **Why ML helps** | 30,000 pages with mixed signals — too complex for simple thresholds. Model catches ~74% vs ~24% for rules. |

```python
# Download the data file directly from the original template repo
!wget https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv -O content_refresh_anonymized.csv

import pandas as pd

df = pd.read_csv('content_refresh_anonymized.csv')

print(f"1. Total pages: {len(df):,}")
declining = (df['trend_direction'] == 'down').sum()
print(f"2. Declining pages: {declining:,} ({declining/len(df)*100:.1f}%)")
visible = df[df['impressions_90d'] > 0]
print(f"3. Pages with impressions: {len(visible):,} ({len(visible)/len(df)*100:.1f}%)")
high_vol = df[df['impressions_90d'] >= 500]
print(f"4. High-volume pages (500+ impressions): {len(high_vol):,}")
avg_pos = df[df['avg_position'].notna() & (df['avg_position'] > 0)]['avg_position'].mean()
print(f"5. Average position: {avg_pos:.1f}")
```

**What these numbers show:**
- All 30,000 pages have search impressions — the starter dataset is already filtered to visible pages.
- 54.2% are marked declining — a large enough pool to build a review queue.
- 16,726 pages have 500+ impressions — high-volume candidates worth prioritizing.
- Average position is 17.0, meaning many pages are mid-to-lower SERP and could move with effort.
- The starter model beats rules 3× — stronger signals likely exist in the full warehouse.

## 4. Careful words

**I can claim:**
- Observable signals (impressions, clicks, position, age, engagement) are associated with the "declining" proxy label in this dataset.
- A learned model can rank pages more accurately than fixed rules on this historical slice.
- The ranked queue helps prioritize limited human review time.

**I cannot claim:**
- Refreshing a page *causes* recovery — this is observational data, not an experiment.
- The model will generalize to all clients or all time periods without re-validation.
- The "declining" label is perfect truth — it's a proxy built from current-window trends, not a future outcome.

**What would make a recommendation wrong:** Seasonality, consolidation (traffic moved to a sibling page), low-volume noise, or a page already being worked on by another team.
